```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("IRIS.csv")
```

```python
# Display the following information about the DataFrame
print("First 10 rows:")
print(df.head(10))

print("Shape:")
print(df.shape)

print("Data types:")
print(df.dtypes)

print("Summary statistics:")
print(df.describe())
```

```python
print("Rows where petal length is greater than 4.5 and species is Iris-virginica:")
print(df[(df["petal_length"] > 4.5) & (df["species"] == "Iris-virginica")])
```

```python
print("Aggregated data by species:")
print(df.groupby("species").agg({
    "sepal_length": ["mean"],
    "petal_width": ["max"],
    "sepal_width": ["std"]
}))
```

```python
print("Petal ratio by species:")
df["petal_ratio"] = df["petal_length"] / df["petal_width"]
print(df.groupby("species")["petal_ratio"].mean())
```

```python
tdf = pd.read_csv("Titanic-Dataset.csv")
column_names = tdf.columns.tolist()
print("Column names: ", end="")
print(", ".join(column_names))

print("\nDisplay only the specified columns: Name, Sex, Age, Fare, Survived")
tdf[["Name", "Sex", "Age", "Fare", "Survived"]].head(10)
```

```python
print("\nPassengers who are female and paid more than 30 for their fare:")
print(tdf[(tdf["Sex"] == "female") & (tdf["Fare"] > 30)])
```

```python
print("\n Group by pClass and compute: survival rate, average age, and average fare:")
print(tdf.groupby("Pclass").agg(
    survival_rate=("Survived", "mean"),
    average_age=("Age", "mean"),
    average_fare=("Fare", "mean")
))
```
