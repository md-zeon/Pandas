# Pandas
Pandas is a powerful and flexible open-source data analysis and manipulation library for Python. It provides data structures like Series and DataFrame, which are designed to handle structured data efficiently. Pandas is widely used in data science, machine learning, and statistical analysis.

```python
import numpy as np
import pandas as pd
```

# Pandas Series
A Pandas Series is a one-dimensional labeled array that can hold data of any type (integer, float, string, etc.). Each element in a Series is associated with an index, which allows for easy access and manipulation of the data.

## How we should visualize the pandas series in our mind
- A Pandas Series can be visualized as a single column of data with an associated index
- The index can be thought of as the "row labels" for the data, while the values represent the actual data points in that column.

# Pandas DataFrame
A Pandas DataFrame is a two-dimensional labeled data structure that can hold data of different types (integer, float, string, etc.) in columns. It is similar to a spreadsheet or SQL table and is one of the most commonly used data structures in Pandas.

## How we should visualize the pandas dataframe in our mind
- A Pandas DataFrame can be visualized as a table with rows and columns, where each column can hold data of a different type. The rows represent individual records, while the columns represent different features or attributes of the data.

```python
# Series in pandas
s = pd.Series([23, 24, 25, 26])
print(s)
print(type(s)) # <class 'pandas.core.series.Series'>

# Indexing
print(s[0])    # 23
print(s[2])    # 25

print(s.index)     # all labels

# Custom Indexing
s2 = pd.Series([23, 24, 25, 26], index = ["Adam", "Eve", "Charlie", "Bob"])
print(s2["Eve"])    # 24
print(s2["Bob"])    # 26

# Vectorized Operations
s1 = pd.Series([1, 2, 3])
s2 = pd.Series([4, 5, 6])

print(s1 + s2)

# Mutable Values but immutable size
s = pd.Series([1, 2, 3, 4, 5])
s[0] = 100

print(s)

changed_s = s.drop(1)
print(changed_s)
print(s)
```

## Series Properties
1. Homogeneous
    - All elements in a Series are of the same data type (e.g., all integers, all floats, all strings).
    - If one element is of a different type, Pandas will attempt to upcast the entire Series to a common data type that can accommodate all elements.
    - This ensures that the Series maintains a consistent data type throughout its lifetime.
    - example: If you create a Series with integers and floats, Pandas will upcast the integers to floats to maintain homogeneity. If you create a Series with integers and strings, Pandas will upcast the integers to objects.
2. Vectorized Operations
    - Pandas Series supports vectorized operations, which means that you can perform operations on the entire Series without the need for explicit loops. This allows for efficient and concise code when working with large datasets.
    - For example, you can add two Series together in a single operation, rather than having to iterate through each element individually.
    - This makes it easy to perform complex calculations on large datasets quickly and efficiently.
3. Handles Missing Data with NaN
    - Pandas Series can handle missing data using the special value NaN (Not a Number). This allows for easy identification and handling of missing values in your data.
    - You can use methods like `isna()` and `fillna()` to identify and fill missing values in a Series, respectively.
4. Mutable values, but immutable size
    - The values in a Pandas Series can be modified, but the size of the Series (i.e., the number of elements) is fixed once it is created. This means that you can change the values of existing elements, but you cannot add or remove elements from the Series after it has been created.
    - If we add or remove elements from a Series, a new Series will be created with the updated size, rather than modifying the original Series in place. This ensures that the original Series remains unchanged and can be used for reference or comparison later on.

### Cleaning of data

- Handling missing values
- `isnull() / isna()`: Returns a boolean Series indicating whether each value is missing (NaN).
- `isnull().sum()`: Returns the count of missing values in each column of the DataFrame.
- `dropna()`: Removes rows or columns with missing values from the DataFrame.
- `fillna(value)`: Fills missing values in the DataFrame with a specified value or method (e.g., forward fill, backward fill).
- `ffill()`: Fills missing values using the last valid observation (forward fill).
- `bfill()`: Fills missing values using the next valid observation (backward fill).

```python
# Cleaning of data
raw_data = pd.read_csv("raw_data.csv")

raw_data.isnull()
raw_data.isna()
raw_data.isnull().sum()  # Count of missing values in each column
raw_data.dropna()  # Remove rows with missing values
raw_data.dropna(axis=1)  # Remove columns with missing values
raw_data.fillna(0)  # Fill missing values with 0
cleaned_data = raw_data.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)  # Fill missing values in 'age' column with mean
cleaned_data

raw_data.ffill()  # Forward fill missing values
raw_data.bfill()  # Backward fill missing values
```

```python
# Handle duplicates
raw_data.duplicated() # Returns a boolean Series indicating duplicate rows
unique_data = raw_data.copy()
unique_data.drop_duplicates(inplace=True) # Remove duplicate rows and modify the original DataFrame in place
unique_data
```
